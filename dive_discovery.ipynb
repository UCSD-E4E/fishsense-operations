```python
from pathlib import Path

import pandas as pd
from PIL import ExifTags, Image
```

```python
data_root = Path('//e4e-nas.ucsd.edu/fishsense_data/REEF/data')
```

```python
df = pd.DataFrame()
```

```python
images = data_root.rglob('*.ORF', case_sensitive=False)
```

```python
dives = {img.parent.absolute() for img in images}
```

```python
df['directory'] = [dive.as_posix() for dive in dives]
```

```python
for dirname in df['directory']:
    path = Path(dirname)
    jpgs = path.glob('*.JPG', case_sensitive=False)
    exifs = (Image.open(jpg).getexif() for jpg in jpgs)
    creation_times = [exif.get(ExifTags.Base.DateTimeOriginal) for exif in exifs]
    print(dirname, creation_times)
```

```python
import sqlite3
from pathlib import Path

from PIL import ExifTags, Image
```

```python
con = sqlite3.connect('fs.db')
```

```python
cur = con.cursor()
```

```python
def load_script(path) -> str:
    with open(path, 'r', encoding='utf-8') as handle:
        return handle.read()
```

```python
cur.executescript(load_script('sql/setup_tables.sql'))
```

```python
cur.execute('''
INSERT INTO canonical_dives (path, date, invalid_image, multiple_date, checksum)
SELECT min(path), date, invalid_image, multiple_date, checksum FROM dives GROUP BY checksum
RETURNING path
;''')
dives = cur.fetchall()
dives
```

```python
cur.close()
con.close()
```

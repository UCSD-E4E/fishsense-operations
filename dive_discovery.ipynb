```python
import datetime as dt
from pathlib import Path

import numpy as np
import pandas as pd
from PIL import ExifTags, Image
from tqdm.auto import tqdm

from backend import get_dive_date
```

```python
data_root = Path('//e4e-nas.ucsd.edu/fishsense_data/REEF/data')
```

```python
df = pd.DataFrame()
```

```python
images = data_root.rglob('*.ORF', case_sensitive=False)
```

```python
dives = {img.parent.absolute() for img in images}
```

```python
df['directory'] = [dive.as_posix() for dive in dives]
```

```python
df['date'] = np.nan
df['invalid_image'] = np.nan
df['multiple_date'] = np.nan
```

```python
for dirname in tqdm(df['directory']):
    path = Path(dirname)
    mean_date, invalid_dates, multiple_dates = get_dive_date(path)
    df.at[dirname, 'date'] = mean_date
    df.at[dirname, 'invalid_image'] = invalid_dates
    df.at[dirname, 'multiple_date'] = multiple_dates
```
